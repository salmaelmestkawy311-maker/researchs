## Part 1: Research Answers

### 1. What is data cleaning, and why is it important?
**Data cleaning** (or data cleansing) is the process of detecting, correcting, or removing inaccurate, incomplete, duplicate, improperly formatted, or corrupted data from a dataset. Real-world datasets collected from web forms, user inputs, or database joins are almost always noisy.

**Why it is important:**
- **Accuracy & Model Reliability:** Ensures downstream analyses, visualizations, and machine learning models produce correct results ("Garbage in, garbage out").
- **Operational Efficiency:** Prevents pipeline errors, crashes, and computation delays caused by unexpected data types or missing values.
- **Consistency:** Standardizes formats, temporal units, and column data types across different records or system integrations.

### 2. How do `isnull()` and `isna()` identify missing values?
In Pandas, `isnull()` and `isna()` are functional aliases—they perform the exact same task. They evaluate every element in a DataFrame or Series and return a boolean mask (`True` where values are missing, such as `NaN`, `None`, or `NaT`, and `False` otherwise).

- **Usage Example:** `df.isnull().sum()` or `df.isna().sum()` returns the total count of missing values per column.

### 3. What is the difference between `dropna()` and `fillna()`?
- **`dropna()`:** Removes (deletes) rows or columns containing missing values (`NaN`).
  - *Example:* `df.dropna(subset=['join_date'])` drops rows where `join_date` is missing.
- **`fillna()`:** Replaces (imputes) missing values with a designated value or statistical measure (e.g., mean, median, mode, or a placeholder string like `"Unassigned"`).
  - *Example:* `df['age'].fillna(df['age'].median(), inplace=True)` replaces missing ages with the median.

### 4. How does `drop_duplicates()` work?
`drop_duplicates()` scans a DataFrame for duplicate rows across all or specified columns and removes repetitive records.
- By default, it keeps the **first** occurrence (`keep='first'`).
- Options include keeping the **last** occurrence (`keep='last'`) or removing **all** duplicate instances (`keep=False`).
- *Example:* `df.drop_duplicates()` retains only unique records.

### 5. How can Pandas convert columns to appropriate data types using `astype()` and `to_datetime()`?
- **`astype()`:** Used for general type conversion (e.g., converting float numbers to integers, string representations to numbers, or objects to categorical types).
  - *Example:* `df['age'] = df['age'].astype(int)`
- **`to_datetime()`:** A specialized utility that parses date/time strings or timestamps into standard Pandas `datetime64` objects, handling multiple temporal formats seamlessly (e.g., passing `format='mixed'`).
  - *Example:* `df['join_date'] = pd.to_datetime(df['join_date'], format='mixed')`

## Part 2: Practical Application

Below is the complete step-by-step practical implementation using Python and Pandas.

```python
import pandas as pd
import numpy as np

# 1 & 2. Create a dataset with 22 records containing missing values, duplicates, and improper data types
raw_data = {
    'employee_id': ['E101', 'E102', 'E103', 'E104', 'E105', 'E106', 'E107', 'E108', 'E109', 'E110', 
                   'E101', 'E111', 'E112', 'E113', 'E114', 'E115', 'E116', 'E117', 'E118', 'E119', 'E120', 'E120'],
    'full_name': ['Alice Smith', 'Bob Jones', 'Charlie Brown', 'Diana Prince', 'Evan Wright', 
                 'Fiona Gallagher', 'George Clark', 'Hannah Abbott', 'Ian Malcolm', 'Julia Roberts', 
                 'Alice Smith', 'Kevin Bacon', 'Laura Croft', 'Mike Ross', 'Nina Williams', 
                 'Oscar Isaac', 'Paul Atreides', 'Quinn Fabray', 'Rachel Green', 'Steve Rogers', 
                 'Tony Stark', 'Tony Stark'],
    'age': ['28', '34', np.nan, '42', '29', '31', '45', np.nan, '50', '36', 
            '28', '38', '27', '30', '33', np.nan, '24', '29', '32', '101', '48', '48'],
    'join_date': ['2021-01-15', '2020/03/10', '2019-11-23', '2022-05-01', '2021-08-19', 
                  '2018-04-12', '2020-09-30', '2022-01-10', '2017-06-15', '2021-12-01', 
                  '2021-01-15', '2019-07-22', '2022-03-14', np.nan, '2020-10-05', 
                  '2021-06-18', '2023-01-01', '2019-12-15', '2022-04-20', '2018-11-11', '2015-05-04', '2015-05-04'],
    'salary': [70000, 85000, 62000, np.nan, 58000, 92000, 75000, 50000, np.nan, 88000, 
               70000, 95000, 68000, 72000, 61000, 105000, 54000, 67000, 78000, 110000, 150000, 150000],
    'department': ['HR', 'IT', 'Finance', 'IT', np.nan, 'Finance', 'HR', 'IT', 'Marketing', np.nan, 
                   'HR', 'IT', 'Marketing', 'Finance', 'HR', 'IT', 'Marketing', 'Finance', 'HR', 'Management', 'Management', 'Management']
}

df_raw = pd.DataFrame(raw_data)

# Save original dataset to CSV
df_raw.to_csv('original_dataset.csv', index=False)

print("--- Original Dataset Head ---")
print(df_raw.head(10))
print("\n--- Original Dataset Info ---")
print(df_raw.info())
```

```python
# 3. Detect and count missing values in each column
print("--- Missing Values Count Per Column ---")
print(df_raw.isnull().sum())

print("\n--- Total Duplicate Records ---")
print(df_raw.duplicated().sum())
```

```python
df_clean = df_raw.copy()

# 4. Remove duplicate records
df_clean = df_clean.drop_duplicates()

# 6. Convert columns to appropriate data types
# Convert age string to numeric float
df_clean['age'] = pd.to_numeric(df_clean['age'])

# Convert join_date to datetime type
df_clean['join_date'] = pd.to_datetime(df_clean['join_date'], format='mixed')

# 5. Handle missing values using suitable strategies
# Fill age NaNs with median age (resilient against extreme outliers like 101)
df_clean['age'] = df_clean['age'].fillna(df_clean['age'].median()).astype(int)

# Fill salary NaNs with overall median salary
df_clean['salary'] = df_clean['salary'].fillna(df_clean['salary'].median())

# Categorical missing value imputation
df_clean['department'] = df_clean['department'].fillna('Unassigned')

# Drop rows where join_date is missing (essential field)
df_clean = df_clean.dropna(subset=['join_date'])

# Save cleaned dataset to CSV
df_clean.to_csv('cleaned_dataset.csv', index=False)

print("--- Cleaning Successful ---")
print(f"Cleaned dataset contains {len(df_clean)} records.")
```

### Short Explanation of Cleaning Decisions

1. **Duplicate Removal:** Used `drop_duplicates()` to remove exact identical rows (e.g., duplicate entries for `E101` and `E120`).
2. **Age & Salary Imputation:** Replaced missing values using **median** values (`fillna()`) rather than mean to avoid bias introduced by outliers (e.g., `age=101`).
3. **Department Imputation:** Imputed missing values with `'Unassigned'` to retain the employee entry without making false departmental assertions.
4. **Dropping Missing `join_date`:** Dropped rows missing `join_date` via `dropna(subset=['join_date'])` because tenure/date calculations require complete records.
5. **Data Type Casting:** Standardized numeric types using `pd.to_numeric()`/`.astype(int)` and temporal data using `pd.to_datetime(..., format='mixed')`.

```python
# 7. Compare the dataset before and after cleaning
print("=================== DATASET COMPARISON ===================")
print(f"Original Shape: {df_raw.shape}")
print(f"Cleaned Shape:  {df_clean.shape}")

print("\n--- Missing Values Before ---")
print(df_raw.isnull().sum())

print("\n--- Missing Values After ---")
print(df_clean.isnull().sum())

print("\n--- Data Types Before ---")
print(df_raw.dtypes)

print("\n--- Data Types After ---")
print(df_clean.dtypes)

print("\n--- Cleaned Dataset Sample ---")
print(df_clean.head())
```
