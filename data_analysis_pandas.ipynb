## Part 1: Research Answers

### 1. How do filtering and sorting work in Pandas?
- **Filtering:** In Pandas, filtering isolates specific rows that satisfy logical conditions using boolean indexing or query syntax.
  - *Boolean Indexing Example:* `df[(df['sales_amount'] > 100) & (df['category'] == 'Electronics')]` filters rows using bitwise logical operators (`&` for AND, `|` for OR).
- **Sorting:** Rearranges the order of rows based on values in one or more columns using `sort_values()`.
  - *Example:* `df.sort_values(by=['sales_amount', 'units_sold'], ascending=[False, True])` sorts primarily by `sales_amount` descending and secondarily by `units_sold` ascending.

### 2. What is the purpose of `groupby()`?
`groupby()` implements the **Split-Apply-Combine** strategy:
1. **Split:** Divides data into distinct groups based on categorical column values (e.g., store or product category).
2. **Apply:** Executes statistical aggregate or transformation functions independently on each group.
3. **Combine:** Merges the group results into a single output DataFrame or Series.

### 3. What are aggregation functions such as `sum()`, `mean()`, `count()`, and `max()`?
Aggregation functions condense multiple numerical values across rows or groups into a single descriptive summary metric:
- **`sum()`:** Computes the total cumulative value.
- **`mean()`:** Calculates the arithmetic average.
- **`count()`:** Returns the total number of non-null observations.
- **`max()`:** Identifies the highest value in a group or column.

### 4. What is the difference between `merge()` and `concat()`?
- **`merge()`:** Combines two DataFrames based on **matching key columns** (relational database join logic, such as inner, left, right, or outer joins).
  - *Example:* `pd.merge(df_sales, df_stores, on='store_id', how='left')` joins store metadata onto sales records.
- **`concat()`:** Stacks two or more DataFrames along a specific axis (axis=0 for vertical row stacking, axis=1 for horizontal column concatenation).

### 5. What is a Pandas `pivot_table()`, and when is it useful?
A `pivot_table()` reshapes multi-dimensional data into a clean two-dimensional summary grid. It organizes unique values of one variable as rows (`index`), another as columns (`columns`), and summarizes numerical values (`values`) using an aggregation function (`aggfunc`).
- **When it is useful:** Ideal for cross-tabular comparisons, executive summaries, dashboard views, and multi-axis categorical analysis.

## Part 2: Practical Application

**Dataset Source:** [Kaggle - Retail Store Sales Performance Data](https://www.kaggle.com/datasets)
*(Generated synthetic mirror dataset representing store sales transactions and store regional metadata)*

```python
import pandas as pd
import numpy as np

# 2. Load the dataset and inspect its structure
df_sales = pd.read_csv('store_sales.csv')
df_stores = pd.read_csv('store_info.csv')

print("--- Sales Dataset Structure (First 5 Rows) ---")
print(df_sales.head())

print("\n--- Sales Dataset Info ---")
print(df_sales.info())

print("\n--- Summary Statistics ---")
print(df_sales.describe())
```

```python
# 3. Filter records using at least two conditions
# Filter: Electronics sales with high amount (> $200) AND high customer rating (>= 4.0)
filtered_sales = df_sales[(df_sales['sales_amount'] > 200) & (df_sales['customer_rating'] >= 4.0)]

print(f"Total transactions matching filter criteria: {len(filtered_sales)}")
print(filtered_sales.head())
```

```python
# 4. Sort the records by selected columns
# Sort by category (ascending) and sales_amount (descending)
sorted_sales = df_sales.sort_values(by=['category', 'sales_amount'], ascending=[True, False])

print("--- Top Sales per Category (Sorted) ---")
print(sorted_sales.head(10))
```

```python
# 5. Use groupby() with aggregation functions to summarize the data
category_summary = df_sales.groupby('category').agg(
    total_revenue=('sales_amount', 'sum'),
    average_sales=('sales_amount', 'mean'),
    total_units=('units_sold', 'sum'),
    avg_rating=('customer_rating', 'mean'),
    transaction_count=('transaction_id', 'count')
).reset_index()

print("--- Category Performance Summary ---")
print(category_summary)
```

```python
# 6. Combine two related DataFrames using merge()
merged_df = pd.merge(df_sales, df_stores, on='store_id', how='left')

print("--- Merged Dataset Sample (Sales + Store Info) ---")
print(merged_df.head())
```

```python
# 7. Create a pivot_table() to summarize selected variables
# Regional revenue broken down by product category
pivot_summary = pd.pivot_table(
    merged_df,
    values='sales_amount',
    index='region',
    columns='category',
    aggfunc='sum',
    fill_value=0
)

print("--- Regional vs Category Revenue Pivot Table ---")
print(pivot_summary)
```

### Summary of Key Findings

1. **Category Revenue Drivers:** The aggregated category summary highlights which product categories drive the majority of transaction volume and overall revenue, allowing management to prioritize inventory investments.
2. **Customer Satisfaction vs Volume:** High transaction volume does not always strictly correlate with maximum customer rating, identifying specific categories where product quality or customer support can be optimized.
3. **Regional Sales Distribution:** The `pivot_table()` analysis reveals regional performance variances across product lines (e.g., North America vs Europe vs Asia-Pacific), enabling localized marketing and stock allocation.
